### `TP01.3. Asterisks`

## `A`

```python
for i in range (6):
    if i == 0 or i == 5:
        star = '*' * 20
        print (star)
    else:
        star = '*                  *'
        print (star)
```

### `B`

```python
for i in range(1,6):
    for j in range(5-i):
        print(" ",end=" ")
    for j in range(2*i-1):
        print("*",end=" ")
    print()
```

### `C`

```python
for i in range (1, 6):
    star = '*' * (2 * i - 1)
    print (star)
```

### `D`

```python
for i in range(1, 10, 2):
    space = 9 - i
    print(" " * space + "*" * i)
```

### `E`

```python
for i in range (1, 6):
    if i == 1 or i == 5:
        space = 4
        print (" " * space + '*')
    elif i == 2 or i == 4:
        space = 2
        print (" " * space + '***')
    else:
        print ('*****')
```
